```python
import json
purchases = dict()

with open("D:\\purchase_log.txt", "r", encoding='utf-8-sig') as file:
    for line in file:
        new_dict = json.loads(line)
        purchases[new_dict['user_id']] = new_dict['category']
```

```python
import csv
with open("D:\\visit_log.csv", 'r', newline = '', encoding='utf-8-sig') as file,\
open("D:\\funnel.csv", 'w', newline='', encoding='utf-8-sig') as csvfile:
    reader = csv.DictReader(file, delimiter=';')
    writer = csv.DictWriter(
                            csvfile,
                            fieldnames=['user_id', 'source', 'category'],
                            delimiter=';',
                            restval='')
    writer.writeheader()
    for row in reader:
        if row['user_id'] in purchases:
            row['category'] = purchases[row['user_id']]
            writer.writerow(row)
```
