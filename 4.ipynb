```python
from datetime import datetime, timedelta

dt = 'The Moscow Times - Wednesday, October 2, 2002'.split(' - ')
new_dt = datetime.strptime(dt[1], '%A, %B %w, %Y').date()
print(new_dt)

dt = 'The Guardian - Friday, 11.10.13'.split(' - ')
new_dt = datetime.strptime(dt[1], '%A, %d.%m.%y').date()
print(new_dt)

dt = 'Daily News - Thursday, 18 August 1977'.split(' - ')
new_dt = datetime.strptime(dt[1], '%A, %d %B %Y').date()
print(new_dt)
```

```python
def date_range(start_date, end_date):
    date_list = []
    try:
        sd = datetime.strptime(start_date, '%Y-%m-%d').date()
        ed = datetime.strptime(end_date, '%Y-%m-%d').date()
    except ValueError:
        pass
    else:
        while sd <= ed:
            date_list.append(sd.strftime('%Y-%m-%d'))
            sd += timedelta(days=1)
    finally:
        print(date_list)

date_range('2026-12-10', '2026-12-12')
```
