```python
def sum_distance(fr, to):
    sum=0
    if fr > to:
        fr, to = to, fr
    while fr <= to:
        sum += fr
        fr += 1
    print(sum)
    
sum_distance(10,5)
```

```python
def trim_and_repeat(str_inp:str, offset:int, repetitions:int):
    for rep in range(repetitions):
        str_inp = str_inp[offset::]
    print(str_inp)

trim_and_repeat('wdecwecwecwec', 3, 2)
```
